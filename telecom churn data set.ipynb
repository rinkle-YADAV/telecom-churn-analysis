```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r"C:\Users\Hp\OneDrive\Desktop\WA_Fn-UseC_-Telco-Customer-Churn.csv")

print(df.shape)          # (7043, 21)
print(df.head())
print(df.info())
print(df['Churn'].value_counts())
print(df['Churn'].value_counts(normalize=True) * 100)

# Visualize churn distribution
df['Churn'].value_counts().plot(kind='bar',
    color=['#1D9E75','#D85A30'], figsize=(6,4))
plt.title('Churn Distribution')
plt.xticks(rotation=0)
plt.savefig('churn_distribution.png', dpi=150)
plt.show()

# Contract type vs churn
sns.countplot(data=df, x='Contract', hue='Churn',
              palette=['#1D9E75','#D85A30'])
plt.title('Churn by Contract Type')
plt.savefig('churn_by_contract.png', dpi=150)
plt.show()
```
